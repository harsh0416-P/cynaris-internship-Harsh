```python
import pandas as pd
```

```python
customers = pd.DataFrame({
    "customer_id": [1, 2, 3, 4],
    "name": ["Harsh", "Rahul", "Priya", "Amit"],
    "region": ["South", "West", "North", "South"]
})

customers 
```

```python
sales = pd.DataFrame({
    "customer_id": [1, 2, 3, 5],
    "product": ["Laptop", "Phone", "Mouse", "Keyboard"],
    "amount": [55000, 25000, 1000, 3000]
})

sales
```

```python
inner_join = pd.merge(
    customers,
    sales,
    on="customer_id",
    how="inner"
)

inner_join 
```

```python
left_join = pd.merge(
    customers,
    sales,
    on="customer_id",
    how="left"
)

left_join
```

```python
right_join = pd.merge(
    customers,
    sales,
    on="customer_id",
    how="right"
)

right_join 
```

```python
outer_join = pd.merge(
    customers,
    sales,
    on="customer_id",
    how="outer"
)

outer_join
```

```python
sales_data = pd.DataFrame({
    "region": ["South", "South", "West", "West", "North", "North"],
    "product": ["Laptop", "Phone", "Laptop", "Phone", "Laptop", "Phone"],
    "amount": [55000, 25000, 60000, 30000, 50000, 20000]
})

sales_data
```

```python
pivot_table = pd.pivot_table(
    sales_data,
    values="amount",
    index="region",
    columns="product",
    aggfunc="sum"
)

pivot_table
```

```python
employees = pd.DataFrame({
    "name": ["Harsh", "Rahul", "Priya", "Amit"],
    "department": ["IT", "Sales", "IT", "Sales"],
    "salary": [50000, 40000, 60000, 35000]
})

employees 
```

```python
result = (
    employees
    .query("salary >= 40000")
    .sort_values("salary", ascending=False)
    .reset_index(drop=True)
)

result
```
