```python

import numpy as np

# Task 1: Create 1D, 2D, and 3D arrays

# One-dimensional array
arr_1d = np.array([10, 20, 30, 40, 50])

# Two-dimensional array
arr_2d = np.array([
    [10, 20, 30],
    [40, 50, 60]
])

# Three-dimensional array
arr_3d = np.array([
    [[1, 2], [3, 4]],
    [[5, 6], [7, 8]]
])

print("1D Array:")
print(arr_1d)
print("Shape:", arr_1d.shape)

print("\n2D Array:")
print(arr_2d)
print("Shape:", arr_2d.shape)

print("\n3D Array:")
print(arr_3d)
print("Shape:", arr_3d.shape)

# Array manipulation
print("\nFirst three values:", arr_1d[:3])
print("Reshaped array:")
print(arr_1d.reshape(5, 1))
```

```python

import numpy as np

# Task 2: Vectorised operations

# Create two arrays
sales_january = np.array([100, 200, 300, 400, 500])
sales_february = np.array([150, 250, 350, 450, 550])

# Vectorised addition
total_sales = sales_january + sales_february
print("Combined sales:", total_sales)

# Vectorised multiplication
doubled_sales = sales_january * 2
print("Doubled January sales:", doubled_sales)

# Aggregation
print("Total January sales:", np.sum(sales_january))
print("Average January sales:", np.mean(sales_january))
print("Maximum January sales:", np.max(sales_january))
print("Minimum January sales:", np.min(sales_january))

# Aggregation across axes in a 2D array
sales = np.array([
    [100, 200, 300],
    [400, 500, 600]
])

print("\nSales by row:", np.sum(sales, axis=1))
print("Sales by column:", np.sum(sales, axis=0))
```

```python

import numpy as np

# Task 3: Boolean masking and fancy indexing

# Sample student marks
marks = np.array([45, 78, 92, 56, 88, 34, 67, 95, 72, 40])

# Boolean masking: select marks greater than or equal to 70
passed_marks = marks[marks >= 70]
print("Marks greater than or equal to 70:", passed_marks)

# Select marks below 40
low_marks = marks[marks < 40]
print("Marks below 40:", low_marks)

# Combine conditions: marks between 50 and 80
middle_marks = marks[(marks >= 50) & (marks <= 80)]
print("Marks between 50 and 80:", middle_marks)

# Fancy indexing: select specific positions
selected_marks = marks[[0, 2, 4, 7]]
print("Marks at selected positions:", selected_marks)

# Update values using a boolean mask
updated_marks = marks.copy()
updated_marks[updated_marks < 40] = 40
print("Marks after replacing values below 40:", updated_marks)
```

```python

# Task 4: Statistical analysis using NumPy
import numpy as np

sales = np.array([120, 150, 180, 200, 220, 250, 280, 300, 350, 400])

print("Sales:", sales)
print("Mean:", np.mean(sales))
print("Median:", np.median(sales))
print("Standard deviation:", np.std(sales))
print("25th percentile:", np.percentile(sales, 25))
print("50th percentile:", np.percentile(sales, 50))
print("75th percentile:", np.percentile(sales, 75))
print("Minimum:", np.min(sales))
print("Maximum:", np.max(sales))

# Correlation between advertising spend and sales
advertising = np.array([10, 15, 20, 25, 30, 35])
revenue = np.array([100, 140, 180, 210, 260, 300])

correlation = np.corrcoef(advertising, revenue)[0, 1]
print("\nAdvertising-sales correlation:", round(correlation, 3))
```

```python

# Task 5: Python loop vs NumPy performance
import numpy as np
import time

n = 1_000_000

# Generate one million values
numbers = list(range(n))
array = np.arange(n)

# Python loop benchmark
start = time.perf_counter()

python_result = [x * 2 for x in numbers]

python_time = time.perf_counter() - start

# NumPy vectorised benchmark
start = time.perf_counter()

numpy_result = array * 2

numpy_time = time.perf_counter() - start

print("Number of values:", n)
print(f"Python loop time: {python_time:.6f} seconds")
print(f"NumPy vectorised time: {numpy_time:.6f} seconds")

if numpy_time > 0:
    print(f"Speed ratio: {python_time / numpy_time:.2f}x")

# Confirm that both approaches produce the same result
print("Results match:", np.array_equal(python_result, numpy_result))
```

```python

# Task 6: NumPy broadcasting
import numpy as np

# Sales of three products across two months
sales = np.array([
    [100, 200, 300],
    [150, 250, 350]
])

# Add a different target to each product column
targets = np.array([10, 20, 30])

adjusted_sales = sales + targets

print("Original sales:")
print(sales)

print("\nTargets:")
print(targets)

print("\nSales after adding targets:")
print(adjusted_sales)
```
