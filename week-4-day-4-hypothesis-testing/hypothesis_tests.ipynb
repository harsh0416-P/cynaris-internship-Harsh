```python
scenarios = {
    "Scenario 1: Delivery Time": {
        "H0": "Average delivery time is 30 minutes.",
        "H1": "Average delivery time is different from 30 minutes."
    },
    "Scenario 2: Website Conversion": {
        "H0": "The new website has the same conversion rate as the old website.",
        "H1": "The new website has a different conversion rate."
    },
    "Scenario 3: Employee Productivity": {
        "H0": "Average productivity is equal in both teams.",
        "H1": "Average productivity differs between the two teams."
    }
}

for scenario, hypotheses in scenarios.items():
    print("\n", scenario)
    print("Null hypothesis (H0):", hypotheses["H0"])
    print("Alternative hypothesis (H1):", hypotheses["H1"])
```

```python
delivery_times = [28, 32, 29, 31, 27, 35, 30, 29, 33, 26]

target = 30

t_stat, p_value = stats.ttest_1samp(delivery_times, popmean=target)

print("Sample mean:", round(np.mean(delivery_times), 2))
print("T-statistic:", round(t_stat, 3))
print("P-value:", round(p_value, 4))

if p_value < 0.05:
    print("Reject H0: Average delivery time differs significantly from 30 minutes.")
else:
    print("Fail to reject H0: Insufficient evidence that average delivery time differs from 30 minutes.")
```

```python
campaign_A = [120, 135, 128, 140, 125, 132, 138, 130]
campaign_B = [145, 150, 142, 155, 148, 152, 147, 158]

t_stat, p_value = stats.ttest_ind(
    campaign_A,
    campaign_B,
    equal_var=False
)

print("Campaign A mean:", round(np.mean(campaign_A), 2))
print("Campaign B mean:", round(np.mean(campaign_B), 2))
print("T-statistic:", round(t_stat, 3))
print("P-value:", round(p_value, 4))

if p_value < 0.05:
    print("Reject H0: The campaign means differ significantly.")
else:
    print("Fail to reject H0: Insufficient evidence of a difference.")
```

```python
def confidence_interval(data, confidence=0.95):
    data = np.array(data)
    n = len(data)
    mean = np.mean(data)
    standard_error = stats.sem(data)
    lower, upper = stats.t.interval(
        confidence,
        df=n - 1,
        loc=mean,
        scale=standard_error
    )
    return mean, lower, upper

mean_delivery, lower_delivery, upper_delivery = confidence_interval(delivery_times)

print("Delivery time mean:", round(mean_delivery, 2))
print("95% confidence interval:",
      (round(lower_delivery, 2), round(upper_delivery, 2)))
```

```python
plt.figure(figsize=(8, 5))

plt.hist(delivery_times, bins=6, edgecolor="black")
plt.axvline(30, color="red", linestyle="--", label="30-minute target")

plt.xlabel("Delivery Time (minutes)")
plt.ylabel("Frequency")
plt.title("Delivery Time Distribution")
plt.legend()
plt.show()
```
