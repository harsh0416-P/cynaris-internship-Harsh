```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings

warnings.filterwarnings("ignore")

plt.style.use("ggplot")
sns.set_palette("deep")

print("Libraries imported successfully.")
```

```python
print("Dataset Information:")
df.info()

print("\nColumn Names:")
print(df.columns.tolist())

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:", df.duplicated().sum())

print("\nStatistical Summary:")
display(df.describe(include="all"))
```

```python
df.columns = df.columns.str.strip()

df = df.drop_duplicates()

df["Order Date"] = pd.to_datetime(
    df["Order Date"], errors="coerce"
)

df["Ship Date"] = pd.to_datetime(
    df["Ship Date"], errors="coerce"
)

df["Sales"] = pd.to_numeric(
    df["Sales"], errors="coerce"
)

df = df.dropna(subset=["Order Date", "Sales"])

df = df[df["Sales"] >= 0]

df = df.sort_values("Order Date").reset_index(drop=True)

print("Data cleaning completed.")
print("Cleaned Shape:", df.shape)
print("Missing Values:", df.isnull().sum().sum())

display(df.head())
```

```python
total_sales = df["Sales"].sum()
total_orders = df["Order ID"].nunique()
total_customers = df["Customer ID"].nunique()
total_products = df["Product ID"].nunique()
average_sales = df["Sales"].mean()

print("TOTAL SALES:", round(total_sales, 2))
print("TOTAL ORDERS:", total_orders)
print("TOTAL CUSTOMERS:", total_customers)
print("TOTAL PRODUCTS:", total_products)
print("AVERAGE TRANSACTION SALES:", round(average_sales, 2))
```

```python
df.columns = df.columns.str.strip()

df["Order Date"] = pd.to_datetime(
    df["Order Date"], errors="coerce"
)

df["Year"] = df["Order Date"].dt.year

print("Year column created successfully.")
display(df[["Order Date", "Year", "Sales"]].head())
```

```python
yearly_sales = df.groupby("Year")["Sales"].sum()

plt.figure(figsize=(10, 5))
yearly_sales.plot(kind="bar")

plt.title("Yearly Sales Performance")
plt.xlabel("Year")
plt.ylabel("Total Sales")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()

display(yearly_sales)
```

```python
category_sales = df.groupby("Category")["Sales"].sum().sort_values(
    ascending=False
)

plt.figure(figsize=(9, 5))
category_sales.plot(kind="bar")

plt.title("Sales by Product Category")
plt.xlabel("Category")
plt.ylabel("Total Sales")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()

display(category_sales)
```

```python
region_sales = df.groupby("Region")["Sales"].sum().sort_values(
    ascending=False
)

plt.figure(figsize=(9, 5))
region_sales.plot(kind="bar")

plt.title("Sales by Region")
plt.xlabel("Region")
plt.ylabel("Total Sales")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()

display(region_sales)
```

```python
monthly_sales = (
    df.set_index("Order Date")["Sales"]
    .resample("MS")
    .sum()
    .reset_index()
)

monthly_sales.columns = ["Date", "Sales"]

print("Monthly Sales Data:")
print("Total Months:", len(monthly_sales))

display(monthly_sales.head(10))
```

```python
plt.figure(figsize=(15, 6))

plt.plot(
    monthly_sales["Date"],
    monthly_sales["Sales"],
    marker="o"
)

plt.title("Monthly Sales Trend")
plt.xlabel("Date")
plt.ylabel("Sales")
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
forecast_df = monthly_sales.rename(
    columns={"Date": "ds", "Sales": "y"}
)

forecast_df = forecast_df.sort_values("ds").reset_index(drop=True)

print("Forecasting Dataset:")
display(forecast_df.head())

print("Forecasting Shape:", forecast_df.shape)
```

```python
split_index = int(len(forecast_df) * 0.80)

train = forecast_df.iloc[:split_index].copy()
test = forecast_df.iloc[split_index:].copy()

print("Training Records:", len(train))
print("Testing Records:", len(test))

print("Training Period:", train["ds"].min(), "to", train["ds"].max())
print("Testing Period:", test["ds"].min(), "to", test["ds"].max())
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error
from math import sqrt

baseline_predictions = np.repeat(
    train["y"].iloc[-1],
    len(test)
)

baseline_mae = mean_absolute_error(test["y"], baseline_predictions)

baseline_rmse = sqrt(
    mean_squared_error(test["y"], baseline_predictions)
)

print("BASELINE MODEL")
print("MAE:", round(baseline_mae, 2))
print("RMSE:", round(baseline_rmse, 2))
```

```python
from prophet import Prophet

model = Prophet(
    yearly_seasonality=True,
    weekly_seasonality=False,
    daily_seasonality=False,
    seasonality_mode="additive",
    interval_width=0.95
)

model.fit(train)

print("Prophet model trained successfully.")
```

```python
test_future = test[["ds"]].copy()

test_forecast = model.predict(test_future)

test_results = test[["ds", "y"]].merge(
    test_forecast[["ds", "yhat", "yhat_lower", "yhat_upper"]],
    on="ds"
)

display(test_results)
```

```python
mae = mean_absolute_error(
    test_results["y"],
    test_results["yhat"]
)

rmse = sqrt(
    mean_squared_error(
        test_results["y"],
        test_results["yhat"]
    )
)

mape = (
    np.abs(
        (test_results["y"] - test_results["yhat"])
        / test_results["y"].replace(0, np.nan)
    ).mean()
) * 100

print("PROPHET MODEL PERFORMANCE")
print("-------------------------")
print("MAE:", round(mae, 2))
print("RMSE:", round(rmse, 2))
print("MAPE:", round(mape, 2), "%")

print("\nBASELINE COMPARISON")
print("Baseline MAE:", round(baseline_mae, 2))
print("Prophet MAE:", round(mae, 2))
```

```python
plt.figure(figsize=(14, 6))

plt.plot(
    test_results["ds"],
    test_results["y"],
    marker="o",
    label="Actual Sales"
)

plt.plot(
    test_results["ds"],
    test_results["yhat"],
    marker="o",
    label="Predicted Sales"
)

plt.title("Actual vs Predicted Monthly Sales")
plt.xlabel("Date")
plt.ylabel("Sales")
plt.legend()
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
model.plot_components(test_forecast)
plt.show()
```

```python
final_model = Prophet(
    yearly_seasonality=True,
    weekly_seasonality=False,
    daily_seasonality=False,
    seasonality_mode="additive",
    interval_width=0.95
)

final_model.fit(forecast_df)

print("Final model trained on complete historical data.")
```

```python
future = final_model.make_future_dataframe(
    periods=12,
    freq="MS"
)

future_forecast = final_model.predict(future)

forecast_output = future_forecast[
    ["ds", "yhat", "yhat_lower", "yhat_upper"]
].tail(12).copy()

forecast_output.columns = [
    "Forecast_Date",
    "Predicted_Sales",
    "Lower_Bound",
    "Upper_Bound"
]

display(forecast_output.round(2))
```

```python
fig = final_model.plot(future_forecast)

plt.title("12-Month Sales Forecast")
plt.xlabel("Date")
plt.ylabel("Sales")

plt.tight_layout()
plt.show()
```

```python
final_model.plot_components(future_forecast)
plt.show()
```

```python
plt.figure(figsize=(15, 6))

plt.plot(
    forecast_df["ds"],
    forecast_df["y"],
    label="Historical Sales",
    marker="o"
)

plt.plot(
    forecast_output["Forecast_Date"],
    forecast_output["Predicted_Sales"],
    label="Forecasted Sales",
    marker="o"
)

plt.fill_between(
    forecast_output["Forecast_Date"],
    forecast_output["Lower_Bound"],
    forecast_output["Upper_Bound"],
    alpha=0.2,
    label="95% Prediction Interval"
)

plt.title("Historical Sales vs Future Sales Forecast")
plt.xlabel("Date")
plt.ylabel("Sales")
plt.legend()
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
best_category = category_sales.idxmax()
best_region = region_sales.idxmax()
best_year = yearly_sales.idxmax()

average_monthly_sales = forecast_df["y"].mean()
forecast_total = forecast_output["Predicted_Sales"].sum()

print("BUSINESS INSIGHTS")
print("=" * 40)

print("Best Performing Category:", best_category)
print("Best Performing Region:", best_region)
print("Highest Sales Year:", best_year)

print(
    "Average Monthly Sales:",
    round(average_monthly_sales, 2)
)

print(
    "Next 12 Months Predicted Sales:",
    round(forecast_total, 2)
)

print(
    "Average Forecasted Monthly Sales:",
    round(forecast_output["Predicted_Sales"].mean(), 2)
)
```
