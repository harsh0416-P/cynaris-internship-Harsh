```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

data = [10, 20, 20, 30, 40, 50, 60, 70, 100]

df = pd.DataFrame(data, columns=["Sales"])
df
```

```python
mean_value = df["Sales"].mean()
print("Mean:", mean_value)
```

```python
median_value = df["Sales"].median()
print("Median:", median_value)
```

```python
mode_value = df["Sales"].mode()
print("Mode:", mode_value.tolist())
```

```python
variance_value = df["Sales"].var()
print("Variance:", variance_value)
```

```python
std_value = df["Sales"].std()
print("Standard Deviation:", std_value)
```

```python
plt.hist(df["Sales"], bins=5, edgecolor="black")

plt.title("Sales Frequency Distribution")
plt.xlabel("Sales")
plt.ylabel("Frequency")

plt.show()
```

```python
q1 = df["Sales"].quantile(0.25)
q2 = df["Sales"].quantile(0.50)
q3 = df["Sales"].quantile(0.75)

print("Q1:", q1)
print("Q2 (Median):", q2)
print("Q3:", q3)
```

```python
p90 = df["Sales"].quantile(0.90)
print("90th Percentile:", p90)
```

```python
group_a = [20, 25, 30, 35, 40, 45, 50]
group_b = [30, 35, 40, 45, 50, 70, 100]

plt.boxplot([group_a, group_b], tick_labels=["Group A", "Group B"])

plt.title("Sales Distribution Comparison")
plt.ylabel("Sales")
plt.show()
```
