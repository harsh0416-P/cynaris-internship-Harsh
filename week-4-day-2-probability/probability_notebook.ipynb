```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
total_customers = 100
purchased = 40
used_discount = 30
both = 20

p_purchase = purchased / total_customers
p_discount = used_discount / total_customers
p_both = both / total_customers

print("P(Purchase):", p_purchase)
print("P(Discount):", p_discount)
print("P(Purchase and Discount):", p_both)
```

```python
table = pd.DataFrame(
    {
        "Purchased": [20, 20],
        "Did Not Purchase": [10, 50]
    },
    index=["Used Discount", "No Discount"]
)

table
```

```python
p_purchase_given_discount = 20 / 30

print("P(Purchase | Discount):", p_purchase_given_discount)
print("Percentage:", p_purchase_given_discount * 100)
```

```python
# Total customers
total = table.values.sum()

# Probability of purchase
p_a = table["Purchased"].sum() / total

# Probability of both purchase and discount
p_a_and_b = table.loc["Used Discount", "Purchased"] / total

# Probability of purchase given discount
p_a_given_b = (
    table.loc["Used Discount", "Purchased"]
    / table.loc["Used Discount"].sum()
)

print("P(A):", p_a)
print("P(A and B):", p_a_and_b)
print("P(A | B):", p_a_given_b)
```

```python
np.random.seed(42)

data = np.random.normal(loc=50, scale=10, size=10000)

plt.hist(data, bins=30, density=True, edgecolor="black")

x = np.linspace(20, 80, 200)
y = (1 / (10 * np.sqrt(2 * np.pi))) * np.exp(
    -0.5 * ((x - 50) / 10) ** 2
)

plt.plot(x, y, linewidth=2)
plt.axvline(40, linestyle="--")
plt.axvline(50, linestyle="--")
plt.axvline(60, linestyle="--")

plt.title("Normal Distribution")
plt.xlabel("Value")
plt.ylabel("Density")
plt.show()
```

```python
np.random.seed(42)

simulations = np.random.binomial(
    n=10,
    p=0.30,
    size=10000
)

values, counts = np.unique(simulations, return_counts=True)

plt.bar(values, counts / len(simulations), edgecolor="black")

plt.title("Binomial Distribution: Sales per 10 Calls")
plt.xlabel("Number of Sales")
plt.ylabel("Probability")
plt.xticks(range(11))
plt.show()

print("Average sales:", simulations.mean())
```

```python
np.random.seed(42)

population = np.random.exponential(scale=10, size=100000)

sample_means = []

for _ in range(5000):
    sample = np.random.choice(population, size=50)
    sample_means.append(sample.mean())

plt.hist(sample_means, bins=40, edgecolor="black")

plt.title("Central Limit Theorem: Distribution of Sample Means")
plt.xlabel("Sample Mean")
plt.ylabel("Frequency")
plt.show()

print("Population mean:", population.mean())
print("Mean of sample means:", np.mean(sample_means))
```
