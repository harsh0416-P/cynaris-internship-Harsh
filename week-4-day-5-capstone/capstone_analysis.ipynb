```python
import os
import glob
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from scipy import stats

sns.set_theme(style="whitegrid")

# Use the KaggleHub download path from your earlier cell
csv_files = glob.glob(os.path.join(path, "**", "*.csv"), recursive=True)

print("CSV files found:", csv_files)

if not csv_files:
    raise FileNotFoundError("No CSV file found in the downloaded dataset.")

df = pd.read_csv(csv_files[0])

print("Dataset loaded successfully!")
print("Rows and columns:", df.shape)
display(df.head())
```

```python
business_question = """
How do sales change over time, which products perform best,
and what business actions could improve sales performance?
"""

print(business_question)
```

```python
print("Dataset shape:", df.shape)

print("\nColumn names:")
print(df.columns.tolist())

print("\nData types and non-null counts:")
df.info()

print("\nMissing values:")
display(df.isnull().sum().to_frame("Missing Values"))

print("\nDuplicate rows:", df.duplicated().sum())

display(df.describe(include="all").T)
```

```python
df_clean = df.copy()

# Remove duplicate records
duplicates_before = df_clean.duplicated().sum()
df_clean = df_clean.drop_duplicates()

print("Duplicate rows removed:", duplicates_before)

# Standardize column names
df_clean.columns = (
    df_clean.columns
    .str.strip()
    .str.lower()
    .str.replace(r"\s+", "_", regex=True)
)

# Convert common date columns if present
for col in df_clean.columns:
    if "date" in col or col.endswith("_time"):
        converted = pd.to_datetime(df_clean[col], errors="coerce")
        if converted.notna().sum() > 0:
            df_clean[col] = converted

print("Cleaned dataset shape:", df_clean.shape)
print("\nMissing values after initial cleaning:")
display(df_clean.isnull().sum().to_frame("Missing Values"))
```

```python
date_candidates = [c for c in df_clean.columns if "date" in c]
sales_candidates = [
    c for c in df_clean.columns
    if c == "sales" or "sales" in c or "revenue" in c
]

print("Possible date columns:", date_candidates)
print("Possible sales columns:", sales_candidates)

if not sales_candidates:
    raise ValueError("No sales/revenue column found. Check the printed column names.")

sales_col = sales_candidates[0]

# Prefer order_date if available
date_col = (
    "order_date" if "order_date" in df_clean.columns
    else date_candidates[0] if date_candidates
    else None
)

df_clean[sales_col] = pd.to_numeric(df_clean[sales_col], errors="coerce")
df_clean = df_clean.dropna(subset=[sales_col])

print("Sales column selected:", sales_col)
print("Date column selected:", date_col)
```

```python
print("Total records:", len(df_clean))
print("Total sales:", round(df_clean[sales_col].sum(), 2))
print("Average sales per record:", round(df_clean[sales_col].mean(), 2))
print("Median sales per record:", round(df_clean[sales_col].median(), 2))
print("Minimum sales:", round(df_clean[sales_col].min(), 2))
print("Maximum sales:", round(df_clean[sales_col].max(), 2))
```

```python
product_cols = [
    c for c in df_clean.columns
    if "product" in c or "category" in c or "sub_category" in c
]

region_cols = [
    c for c in df_clean.columns
    if "region" in c or "state" in c or "city" in c
]

print("Product/category columns:", product_cols)
print("Region/location columns:", region_cols)
```

```python
if product_cols:
    group_col = product_cols[0]
    category_sales = (
        df_clean.groupby(group_col)[sales_col]
        .sum()
        .sort_values(ascending=False)
    )

    display(category_sales.to_frame("Total Sales"))

    category_sales.plot(kind="bar", figsize=(10, 5))
    plt.title(f"Total Sales by {group_col}")
    plt.xlabel(group_col.replace("_", " ").title())
    plt.ylabel("Total Sales")
    plt.xticks(rotation=45, ha="right")
    plt.tight_layout()
    plt.show()
else:
    print("No product/category column found; skipping this analysis.")
```

```python
if region_cols:
    region_col = region_cols[0]
    region_sales = (
        df_clean.groupby(region_col)[sales_col]
        .sum()
        .sort_values(ascending=False)
    )

    display(region_sales.to_frame("Total Sales"))

    region_sales.plot(kind="bar", figsize=(10, 5))
    plt.title(f"Total Sales by {region_col}")
    plt.xlabel(region_col.replace("_", " ").title())
    plt.ylabel("Total Sales")
    plt.xticks(rotation=45, ha="right")
    plt.tight_layout()
    plt.show()
else:
    print("No region/location column found; skipping this analysis.")
```

```python
plt.figure(figsize=(8, 5))
sns.histplot(df_clean[sales_col].dropna(), bins=30, kde=True)
plt.title("Distribution of Sales")
plt.xlabel("Sales")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()
```

```python
if date_col:
    df_clean[date_col] = pd.to_datetime(df_clean[date_col], errors="coerce")
    dated_df = df_clean.dropna(subset=[date_col]).copy()

    dated_df["month"] = dated_df[date_col].dt.to_period("M").dt.to_timestamp()

    monthly_sales = (
        dated_df.groupby("month")[sales_col]
        .sum()
        .sort_index()
    )

    display(monthly_sales.to_frame("Total Sales"))

    monthly_sales.plot(figsize=(12, 5), marker="o")
    plt.title("Monthly Sales Trend")
    plt.xlabel("Month")
    plt.ylabel("Total Sales")
    plt.tight_layout()
    plt.show()
else:
    print("No date column found; skipping time-series analysis.")
```

```python
sales_data = df_clean[sales_col].dropna()

print("Mean:", round(sales_data.mean(), 2))
print("Median:", round(sales_data.median(), 2))
print("Standard deviation:", round(sales_data.std(), 2))
print("25th percentile:", round(sales_data.quantile(0.25), 2))
print("75th percentile:", round(sales_data.quantile(0.75), 2))
```

```python
n = len(sales_data)

if n > 1:
    mean_sales = sales_data.mean()
    standard_error = stats.sem(sales_data)

    lower, upper = stats.t.interval(
        0.95,
        df=n - 1,
        loc=mean_sales,
        scale=standard_error
    )

    print("Average sales:", round(mean_sales, 2))
    print("95% confidence interval:", (round(lower, 2), round(upper, 2)))
else:
    print("Not enough observations to calculate a confidence interval.")
    
```

```python
numeric_df = df_clean.select_dtypes(include="number")

if numeric_df.shape[1] >= 2:
    correlation = numeric_df.corr()

    plt.figure(figsize=(10, 7))
    sns.heatmap(correlation, annot=True, cmap="coolwarm", fmt=".2f")
    plt.title("Numeric Feature Correlation")
    plt.tight_layout()
    plt.show()
else:
    print("At least two numeric columns are required for a correlation heatmap.")
```

```python
print("CAPSTONE BUSINESS FINDINGS")
print("=" * 35)

print(f"Records analysed: {len(df_clean):,}")
print(f"Total sales: {df_clean[sales_col].sum():,.2f}")
print(f"Average sales per record: {df_clean[sales_col].mean():,.2f}")

if product_cols:
    group_col = product_cols[0]
    top_category = (
        df_clean.groupby(group_col)[sales_col]
        .sum()
        .sort_values(ascending=False)
    )

    print(f"\nTop-performing {group_col}: {top_category.index[0]}")
    print(f"Sales: {top_category.iloc[0]:,.2f}")

if region_cols:
    region_col = region_cols[0]
    top_region = (
        df_clean.groupby(region_col)[sales_col]
        .sum()
        .sort_values(ascending=False)
    )

    print(f"\nHighest-sales {region_col}: {top_region.index[0]}")
    print(f"Sales: {top_region.iloc[0]:,.2f}")

print("\nPotential recommendations to investigate:")
print("1. Review high-performing products/categories for growth opportunities.")
print("2. Investigate lower-performing categories or regions before reallocating budget.")
print("3. Use the sales trend to support inventory and promotional planning.")
```

```python
import pandas as pd
import os

df = pd.read_csv("data/sales.csv")

print("Dataset loaded successfully!")
print("Rows and columns:", df.shape)
display(df.head())
```

```python
# Create a copy so the original data stays unchanged
df_clean = df.copy()

# Standardize column names
df_clean.columns = (
    df_clean.columns
    .str.strip()
    .str.lower()
    .str.replace(r"\s+", "_", regex=True)
)

# Check data quality
print("Original shape:", df.shape)

print("\nMissing values:")
display(df_clean.isnull().sum().to_frame("Missing Values"))

print("\nDuplicate rows:", df_clean.duplicated().sum())

# Remove duplicate rows
df_clean = df_clean.drop_duplicates()

# Convert date columns
for col in df_clean.columns:
    if "date" in col:
        df_clean[col] = pd.to_datetime(
            df_clean[col], errors="coerce", dayfirst=True
        )

print("\nCleaned shape:", df_clean.shape)
print("Cleaning completed!")
display(df_clean.head())
```

```python
# 1. Basic sales statistics
sales_col = "sales"

print("Total records:", len(df_clean))
print("Total sales:", round(df_clean[sales_col].sum(), 2))
print("Average sales:", round(df_clean[sales_col].mean(), 2))
print("Median sales:", round(df_clean[sales_col].median(), 2))
print("Highest sale:", round(df_clean[sales_col].max(), 2))
print("Lowest sale:", round(df_clean[sales_col].min(), 2))

# 2. Top 10 products by sales
if "product_name" in df_clean.columns:
    top_products = (
        df_clean.groupby("product_name")["sales"]
        .sum()
        .sort_values(ascending=False)
        .head(10)
    )

    display(top_products.to_frame("Total Sales"))

# 3. Sales by region
if "region" in df_clean.columns:
    region_sales = (
        df_clean.groupby("region")["sales"]
        .sum()
        .sort_values(ascending=False)
    )

    display(region_sales.to_frame("Total Sales"))

print("EDA completed!")
```

```python
print("Dataset shape:", df_clean.shape)
print("Columns:", df_clean.columns.tolist())
display(df_clean.describe(include="all").T)
```

```python
import os
import matplotlib.pyplot as plt
import seaborn as sns

os.makedirs("outputs", exist_ok=True)
sns.set_theme(style="whitegrid")

# 1. Sales distribution
plt.figure(figsize=(8, 5))
sns.histplot(df_clean["sales"].dropna(), bins=30, kde=True)
plt.title("Sales Distribution")
plt.xlabel("Sales")
plt.tight_layout()
plt.savefig("outputs/sales_distribution.png", dpi=150, bbox_inches="tight")
plt.show()

# 2. Top 10 products
if "product_name" in df_clean.columns:
    top_products = df_clean.groupby("product_name")["sales"].sum().nlargest(10)
    plt.figure(figsize=(10, 6))
    top_products.sort_values().plot(kind="barh")
    plt.title("Top 10 Products by Sales")
    plt.xlabel("Total Sales")
    plt.tight_layout()
    plt.savefig("outputs/top_products.png", dpi=150, bbox_inches="tight")
    plt.show()

# 3. Sales by region
if "region" in df_clean.columns:
    region_sales = df_clean.groupby("region")["sales"].sum().sort_values(ascending=False)
    plt.figure(figsize=(8, 5))
    region_sales.plot(kind="bar")
    plt.title("Total Sales by Region")
    plt.ylabel("Total Sales")
    plt.xticks(rotation=30)
    plt.tight_layout()
    plt.savefig("outputs/sales_by_region.png", dpi=150, bbox_inches="tight")
    plt.show()

# 4. Monthly sales trend
if "order_date" in df_clean.columns:
    dates = df_clean.dropna(subset=["order_date"]).copy()
    dates["month"] = dates["order_date"].dt.to_period("M").dt.to_timestamp()
    monthly_sales = dates.groupby("month")["sales"].sum()

    plt.figure(figsize=(12, 5))
    monthly_sales.plot(marker="o")
    plt.title("Monthly Sales Trend")
    plt.ylabel("Total Sales")
    plt.tight_layout()
    plt.savefig("outputs/monthly_sales_trend.png", dpi=150, bbox_inches="tight")
    plt.show()

print("Charts saved in:", os.path.abspath("outputs"))
print("Files:", os.listdir("outputs"))
```

```python
import numpy as np
from scipy import stats

sales = df_clean["sales"].dropna()

print("Mean:", round(sales.mean(), 2))
print("Median:", round(sales.median(), 2))
print("Standard deviation:", round(sales.std(), 2))
print("25th percentile:", round(sales.quantile(0.25), 2))
print("75th percentile:", round(sales.quantile(0.75), 2))

# 95% confidence interval for the mean
if len(sales) > 1:
    mean = sales.mean()
    se = stats.sem(sales)
    lower, upper = stats.t.interval(
        0.95, df=len(sales)-1, loc=mean, scale=se
    )
    print("95% confidence interval:", (round(lower, 2), round(upper, 2)))
```
